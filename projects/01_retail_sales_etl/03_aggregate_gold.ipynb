```python
%sql
CREATE SCHEMA IF NOT EXISTS gold;

CREATE OR REPLACE TABLE workspace.gold.product_revenue
AS
SELECT
    product_id,
    product_name,
    SUM(price * qty)             AS total_revenue,
    SUM(qty)                     AS total_units_sold,
    COUNT(DISTINCT order_number) AS total_orders
FROM silver.sales_clean
GROUP BY product_id, product_name
ORDER BY total_revenue DESC;



CREATE OR REPLACE TABLE workspace.gold.customer_orders
AS
SELECT
    customer_id,
    customer_name,
    COUNT(DISTINCT order_number) AS total_orders,
    SUM(price * qty)             AS total_spent,
    SUM(price * qty) / COUNT(DISTINCT order_number) AS avg_order_value,
    SUM(qty)                     AS total_units
FROM silver.sales_clean
GROUP BY customer_id, customer_name
ORDER BY total_spent DESC;









```

```python
%sql
SELECT * FROM workspace.gold.product_revenue;




SELECT *
FROM (
    SELECT
        product_id,
        product_name,
        total_revenue,
        total_units_sold,
        total_orders,
        DENSE_RANK() OVER (ORDER BY total_revenue DESC) AS revenue_rank
    FROM gold.product_revenue
)
ORDER BY revenue_rank ASC




```
